```python
def numerical_target_analysis(df, feature, target="addicted_label", q=10):

    print(f"\n{'='*60}")
    print(f"Feature: {feature}")

    print("\nTarget group statistics:")

    print(
        df.groupby("addicted_label")[feature]
        .agg(["mean", "median", "std", "min", "max", "count"])
    )

    temp = df[[feature, target]].copy()

    temp["bin"] = pd.qcut(
        temp[feature],
        q=q,
        duplicates="drop"
    )

    print("\nTarget probability by feature bin:")

    print(
        temp.groupby("bin", observed=True)[target]
        .agg(["mean", "count"])
    )
```

```python
target = "addicted_label"

num_cols = (
    df.select_dtypes(include="number")
    .columns
    .drop([target, "id"], errors="ignore")
    .tolist()
)

for col in num_cols:
    numerical_target_analysis(df, col)
```

1. total screnn time = (daily screen time)*5 + weekend screen time - study 
2. non productive hour = social media + game
3. notification per app = notification / app open per day 
4. weekly screen time = daily screen time * 5 + weend screen time *2 
5. screen to productive hour = daily screen time / study hour
6. screen to sleep ratio = daily screen time / skeep hour
7. phone checking intensity = app per day / daily screen time 


```python
for data in [df, df_t]:

    data["total_screen_time"] = (
        data["daily_screen_time_hours"] * 5
        + data["weekend_screen_time"] * 2
        - data["work_study_hours"]
    )

    data["non_productive_hours"] = (
        data["social_media_hours"]
        + data["gaming_hours"]
    )

    data["c"] = (
        data["notifications_per_day"]
        / (data["app_opens_per_day"] + 1e-8)
    )

    data["weekly_screen_time"] = (
        data["daily_screen_time_hours"] * 5
        + data["weekend_screen_time"] * 2
    )

    data["screen_to_productive_ratio"] = (
        data["daily_screen_time_hours"]
        / (data["work_study_hours"] + 1e-8)
    )

    data["screen_to_sleep_ratio"] = (
        data["daily_screen_time_hours"]
        / (data["sleep_hours"] + 1e-8)
    )

    data["phone_checking_intensity"] = (
        data["app_opens_per_day"]
        / (data["daily_screen_time_hours"] + 1e-8)
    )
```
